```python
x = data.drop('selling_price', axis=1)
y = data['selling_price']
```

```python
x
```

```python
y
```

```python
x.drop('car_name', axis=1, inplace=True)
```

```python
x_num = x.select_dtypes(include=['int64', 'float64']).columns
```

```python
from sklearn.model_selection import train_test_split
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
x_train
```

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
```

```python
categorical_columns = [
    'brand',
    'model',
    'seller_type',
    'fuel_type',
    'transmission_type'
]
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(handle_unknown='ignore'), categorical_columns),
        ('num', StandardScaler(), x_num)
    ],
    remainder='passthrough'
)
```

```python
x_train
```

```python
from sklearn.linear_model import LinearRegression

model = LinearRegression()

```

```python
from sklearn.pipeline import Pipeline
pipeline = Pipeline(steps=[
    ('pipeline', preprocessor),
    ('model', model)
])

pipeline.fit(x_train, y_train)
```

```python
y_pred = pipeline.predict(x_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score

print("MAE:", mean_absolute_error(y_test, y_pred))
print("MSE:", mean_squared_error(y_test, y_pred))
print("R2:", r2_score(y_test, y_pred))
```

```python
# lasso and ridge regression
from sklearn.linear_model import Lasso, Ridge
lasso = Lasso(max_iter=10000)
ridge = Ridge(max_iter=10000)
lasso_pipeline = Pipeline(steps=[
    ('pipeline', preprocessor),
    ('model', lasso)
])
ridge_pipeline = Pipeline(steps=[
    ('pipeline', preprocessor),
    ('model', ridge)
])
y_pred_lasso=lasso_pipeline.fit(x_train, y_train).predict(x_test)
y_pred_ridge=ridge_pipeline.fit(x_train, y_train).predict(x_test)
```

```python
import pandas as pd

result = pd.DataFrame({
    'Actual Price': y_test.values,
    'Lasso Prediction': y_pred_lasso,
    'Ridge Prediction': y_pred_ridge
})

print(result.head(10))
```

```python
import pickle
with open('linearmodel.pkl','wb') as f:
    pickle.dump(pipeline,f)

with open('lassomodel.pkl','wb') as f:
    pickle.dump(lasso_pipeline,f)

with open('ridgemodel.pkl','wb') as f: 
    pickle.dump(ridge_pipeline,f)
```
